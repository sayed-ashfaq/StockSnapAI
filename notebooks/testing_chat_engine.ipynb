```python
import os
import tempfile
from pathlib import Path
from datetime import datetime

# Import your own services/utilities
from src.rag_system.document_service import DocumentService
from src.rag_system.vector_service import VectorService
from src.rag_system.chat_engine import ChatService
from utils.files_io import generate_document_id, get_file_type
from config.config import settings
from logger import GLOBAL_LOGGER as logger

# ---------------- Test in Notebook ---------------- #
# Step 1: Initialize services
document_service = DocumentService()
vector_service = VectorService()
chat_service = ChatService(vector_service)

# Step 2: Select a file (change this to your file path)
file_path = "C:\\Users\\302sy\\Desktop\\Generative AI\\StockSnapAI\\data\\4259054b-0c8e-4eaf-950a-56b318b34256.pdf"   # e.g. keep a PDF file in your notebook folder
filename = Path(file_path).name

# Step 3: Validate + read file
with open(file_path, "rb") as f:
    file_content = f.read()

file_type, _ = get_file_type(filename)

# Step 4: Generate document_id
document_id = generate_document_id(filename, file_content)

# Step 5: Process file
documents, metadata = document_service.process_single_file(file_path, file_type)

# Step 6: Store in vector DB
doc_info = {
    "document_id": document_id,
    "filename": filename,
    "file_type": file_type.value,
    "processed_at": datetime.now(),
    "chunk_count": len(documents),
    "has_images": metadata.get("has_images", False),
    "image_count": metadata.get("extracted_images", 0),
}
vector_service.add_documents(document_id, documents)

# Step 7: Generate summary (use first few chunks)
summary_content = "\n\n".join([doc.page_content for doc in documents[:3]])
summary = chat_service.generate_summary(document_id, summary_content)

print("===== SUMMARY =====")
print(summary.get("summary", "No summary"))
print("===================\n")

# Step 8: Interactive chat loop
print("Start chatting with your document (type 'exit' to quit):")
while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        break
    try:
        response = chat_service.chat_with_document(document_id, user_input)
        print(f"Bot: {response}")
    except Exception as e:
        print(f"Error: {e}")
```

```python
# Step 8: Interactive chat loop with history
from src.rag_system.schemas import ChatMessage

chat_history = []   # keep conversation context

print("Start chatting with your document (type 'exit' to quit):")
while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        break

    try:
        # pass history to your chat service
        response = chat_service.chat_with_document(
            document_id=document_id,
            question=user_input,
            chat_history=[ChatMessage(role=msg["role"], content=msg["content"]) for msg in chat_history]
        )

        print(f"Bot: {response.answer}")

        # Append to history as ChatMessages
        chat_history.append({'role': "user", "content": user_input})
        chat_history.append({'role': "assitant", "content": response.answer})


    except Exception as e:
        print(f"Error: {e}")
```
